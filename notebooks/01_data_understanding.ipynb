## Hidden Missing Values Investigation

```python
# Check for blank strings in TotalCharges

(df["TotalCharges"] == "").sum()
```

```python
df[df["TotalCharges"] == ""]
```

```python
# Check for values containing only whitespace

(df["TotalCharges"].str.strip() == "").sum()
```

```python
df["TotalCharges"].dtype
```

```python
df["TotalCharges"].head(10)
```

```python
pd.to_numeric(df["TotalCharges"], errors="coerce").isna().sum()
```

### Observation

The `TotalCharges` column contains **11 hidden missing values** represented as blank whitespace rather than true `NaN` values.

Although `df.isnull()` reported no missing values, converting the column to a numeric data type using `pd.to_numeric(..., errors="coerce")` revealed these hidden missing values.

These records are expected to correspond to customers with very short or zero tenure and will be handled during the Data Cleaning phase.

```python
# Display records with hidden missing values

hidden_missing = df[pd.to_numeric(df["TotalCharges"], errors="coerce").isna()]

hidden_missing
```

```python
hidden_missing[["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]]
```
